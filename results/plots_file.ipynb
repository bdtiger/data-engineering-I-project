```python
#import library for plotting
import matplotlib.pyplot as plt
```

**Horizontal Scaling (5GB Data)**

```python
#horizontal scaling data
# same data size = 5 GB
# workers increase from 1 to 3
workers_h = [1, 2, 3]
runtime_h = [142, 89, 81]

# Step 3: create horizontal scaling runtime plot
plt.figure(figsize=(8,5))
plt.plot(workers_h, runtime_h, marker='o', linewidth=2, label='Runtime')

# labels and title
plt.title('Horizontal Scaling: Runtime vs Number of Workers (5 GB)')
plt.xlabel('Number of Workers')
plt.ylabel('Runtime (seconds)')

# show exact values on points
for x, y in zip(workers_h, runtime_h):
    plt.text(x, y + 2, str(y), ha='center')

# extra formatting
plt.xticks(workers_h)
plt.legend()
#plt.grid(True)

plt.savefig('horizontal_scaling_runtime.png', dpi=300, bbox_inches='tight')

# show plot
plt.show()
```

**Speed Up for Horizontal Scaling**

```python
#calculate speedup for horizontal scaling
# speedup = time with 1 worker / time with N workers
base_time = runtime_h[0]
speedup_h = [base_time / t for t in runtime_h]

# ideal speedup line
ideal_speedup = [1, 2, 3]

# Step 5: create speedup plot
plt.figure(figsize=(8,5))
plt.plot(workers_h, speedup_h, marker='o', linewidth=2, label='Actual Speedup')
plt.plot(workers_h, ideal_speedup, marker='o', linestyle='--', linewidth=2, label='Ideal Speedup')

# labels and title
plt.title('Horizontal Scaling: Speedup vs Number of Workers (5 GB)')
plt.xlabel('Number of Workers')
plt.ylabel('Speedup')

# show exact values on points
for x, y in zip(workers_h, speedup_h):
    plt.text(x, y + 0.05, f'{y:.2f}', ha='center')

# extra formatting
plt.xticks(workers_h)
#plt.grid(True)
plt.legend()

plt.savefig('horizontal_scaling_speedup.png', dpi=300, bbox_inches='tight')

# show plot
plt.show()
```

**Vertical Scaling (with 5GB Data)**

```python
#vertical scaling data for 5 GB
# workers fixed = 3
# cores increase from 1 to 2
cores_v_5gb = [1, 2]
runtime_v_5gb = [83, 81]

# Step 7: create vertical scaling plot for 5 GB
plt.figure(figsize=(8,5))
plt.plot(cores_v_5gb, runtime_v_5gb, marker='o', linewidth=2, label='Runtime')

# labels and title
plt.title('Vertical Scaling: Runtime vs Cores per Executor (5 GB, 3 Workers)')
plt.xlabel('Cores per Executor')
plt.ylabel('Runtime (seconds)')

# show exact values on points
for x, y in zip(cores_v_5gb, runtime_v_5gb):
    plt.text(x, y + 1, str(y), ha='center')

# extra formatting
plt.xticks(cores_v_5gb)
#plt.grid(True)
plt.legend()

plt.savefig('vertical_scaling_5gb.png', dpi=300, bbox_inches='tight')

# show plot
plt.show()
```

**Weak Scaling**

```python
# Step 10: weak scaling data
# workload increases with workers
# current available points are W-1 and W-2
workers_w = [1, 2]
runtime_w = [142, 173]
data_w = [5, 10]

# Step 11: create weak scaling plot
plt.figure(figsize=(8,5))
plt.plot(workers_w, runtime_w, marker='o', linewidth=2, label='Runtime')

# labels and title
plt.title('Weak Scaling: Runtime vs Number of Workers')
plt.xlabel('Number of Workers')
plt.ylabel('Runtime (seconds)')

# show runtime and data size on each point
for x, y, d in zip(workers_w, runtime_w, data_w):
    plt.text(x, y + 3, f'{y}s, {d}GB', ha='center')

# extra formatting
plt.xticks(workers_w)
#plt.grid(True)
plt.legend()

plt.savefig('weak_scaling.png', dpi=300, bbox_inches='tight')

# show plot
plt.show()
```
